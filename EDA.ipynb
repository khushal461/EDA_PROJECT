```python
# Create a function to detect the outliers and apply it on a dataframe
def Outliers(data,col):
  q1 = data[col].quantile(0.25)
  q3 = data[col].quantile(0.75)
  IQR = q3-q1
  lower_fence =q1 -1.5 * (IQR)
  upper_fence =q3 +1.5 * (IQR)
  outlier = data[(data[col]<lower_fence) | (data[col]>upper_fence)]
  return lower_fence,upper_fence,outlier
```

```python
df.columns
```

```python
col = ['Temperature','windspeed','humidity']
for column in col:
  lf,uf,Outl = Outliers(df,column)
  print("column Name:",column)
  print("lower Fence:",lf)
  print("Upper Fence:",uf)
  print("Outliers:",Outl.shape[0])
  print("***************************************")
```
